```python
def print_board(state):
    """Utility to print state in a 3x3 grid format."""
    for i in range(0, 9, 3):
        print(" ".join(map(str, state[i:i+3])))
    print()

def get_neighbors(state):
    """Generates valid next states by moving the blank space (0)."""
    neighbors = []
    zero_idx = state.index(0)
    row, col = divmod(zero_idx, 3)

    # Define possible moves: (row_offset, col_offset, index_change)
    moves = []
    if row > 0: moves.append(-3) # Up
    if row < 2: moves.append(3)  # Down
    if col > 0: moves.append(-1) # Left
    if col < 2: moves.append(1)  # Right

    for move in moves:
        new_zero_idx = zero_idx + move
        state_list = list(state)
        # Swap blank tile with adjacent tile
        state_list[zero_idx], state_list[new_zero_idx] = state_list[new_zero_idx], state_list[zero_idx]
        neighbors.append(tuple(state_list))

    return neighbors

def solve_8_puzzle_dfs(start_state, goal_state, max_depth=30):
    """Solves the 8-puzzle using Depth-First Search with a depth limit."""
    # Stack stores tuples of (current_state, path_taken)
    stack = [(start_state, [start_state])]
    # Map state -> minimum depth reached (prevents redundant deep visits)
    visited = {start_state: 0}

    while stack:
        current_state, path = stack.pop()

        # Goal check
        if current_state == goal_state:
            return path

        # Enforce depth boundary
        if len(path) - 1 >= max_depth:
            continue

        # Explore neighbors
        for neighbor in get_neighbors(current_state):
            depth = len(path)
            if neighbor not in visited or depth < visited[neighbor]:
                visited[neighbor] = depth
                stack.append((neighbor, path + [neighbor]))

    return None

# --- Example Usage ---
if __name__ == "__main__":
    # 0 represents the empty slot
    start_state = (1, 2, 3, 4, 0, 6, 7, 5, 8)
    goal_state  = (1, 2, 3, 4, 5, 6, 7, 8, 0)

    solution_path = solve_8_puzzle_dfs(start_state, goal_state, max_depth=30)

    if solution_path:
        print(f"Solution found in {len(solution_path) - 1} moves:\n")
        for step, state in enumerate(solution_path):
            print(f"Step {step}:")
            print_board(state)
    else:
        print("No solution found within the specified depth limit.")
```
