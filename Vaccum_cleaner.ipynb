```python
class VacuumCleanerAgent:
    def __init__(self):
        self.location = "A"

    def perceive(self, environment):
        return environment[self.location]

    def act(self, environment):
        status = self.perceive(environment)

        if status == "Dirty":
            print(f"Room {self.location} is dirty → Cleaning...")
            environment[self.location] = "Clean"

        elif self.location == "A":
            print("Room A is clean → Moving to Room B")
            self.location = "B"

        else:
            print("Room B is clean → Moving to Room A")


# Environment
environment = {
    "A": "Dirty",
    "B": "Dirty"
}

# Create the agent
agent = VacuumCleanerAgent()

# Run the agent
for step in range(6):
    print(f"\nStep {step + 1}")
    print("Environment:", environment)
    agent.act(environment)

print("\nFinal Environment:", environment)
```
